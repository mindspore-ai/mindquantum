```python
# ========== Fig3(b)(c) 完整成果代码 ==========
import sys
import os
sys.path.insert(0, 'Fig3')

import numpy as np
import matplotlib.pyplot as plt
import mindspore as ms
from mindquantum.core.parameterresolver import PRGenerator

# MindSpore context
ms.set_context(mode=ms.PYNATIVE_MODE, device_target="CPU")

# ========== 方法1: alternating_train (Adam + REINFORCE) ==========
print("=" * 60)
print("方法1: 交替训练 (Adam + REINFORCE)")
print("=" * 60)

from fig3bc_functions import alternating_train_mq, plot_fig3b, plot_fig3c, SHVQEEvaluator, vqe_circuit, _single_qubit_layer, hamiltonian_xxz

# 训练配置
n, depth = 8, 2
print(f"训练配置: n={n}, depth_u={2*depth}")

# 运行训练
hist = alternating_train_mq(
    n_qubits=n, depth=depth, delta=1.0, pbc=True,
    T=800, N=800,  # 完整训练
    tau0=2.0, tau_min=0.25, tau_decay=0.993,
    lr_alpha=0.06, entropy_coef=0.01, baseline_ema=0.06,
    lr_theta=0.03, lr_phi=0.03,
    single_rot_u='rx-ry-rz', single_rot_t='rz-rx-rz',
    seed=2025
)

print("\n训练完成！绘制Fig3(b)和Fig3(c)...")

# 绘制Fig3(b)
plot_fig3b(hist)

# 计算并绘制Fig3(c)
n_val = hist['policy'].L * 2
H = hamiltonian_xxz(n_val, delta=1.0, pbc=True)
U_circ, theta_names = vqe_circuit(n_val, depth, prefix='theta', single_rot='rx-ry-rz', pbc=True)
T1q = _single_qubit_layer(n_val, prefix='phi', layer='T', single_rot='rz-rx-rz', gen=PRGenerator('phi'))
T1q.as_ansatz()
phi_names = list(T1q.ansatz_params_name)
ev = SHVQEEvaluator(n_val, H, U_circ, theta_names, T1q, phi_names)

k = n_val // 2
codes_array = np.array([format(i, f'0{k}b') for i in range(1 << k)])[::-1]
plot_fig3c(ev, codes_array, hist['theta'], hist['phi'])
plt.show()

# ========== 方法2: SPSA方法 ==========
print("\n" + "=" * 60)
print("方法2: SPSA方法 (Adam + SPSA)")
print("=" * 60)

from fig3bc_another_functions import train_with_empirical_log, plot_fig3b_spsa, plot_fig3c_from_results

# 设置评估器
n_spsa = 8
depth_u = 4
rot_u = 'rx-ry-rz'
rot_t = 'rz-ry-rz'
pbc_H = True
pbc_U = True

print(f"SPSA训练配置: n={n_spsa}, depth_u={depth_u}")

U_circ_spsa, theta_names_spsa = vqe_circuit(n_spsa, depth_u, prefix='theta', single_rot=rot_u, pbc=pbc_U)
T1q_spsa = _single_qubit_layer(n_qubits=n_spsa, prefix='phi', layer='T', single_rot=rot_t, gen=PRGenerator('phi'))
T1q_spsa.as_ansatz()
phi_names_spsa = list(T1q_spsa.ansatz_params_name)
H_spsa = hamiltonian_xxz(n_spsa, delta=1.0, pbc=pbc_H)
ev_spsa = SHVQEEvaluator(n_spsa, H_spsa, U_circ_spsa, theta_names_spsa, T1q_spsa, phi_names_spsa)

# 运行SPSA训练
print("运行SPSA训练...")
res_spsa = train_with_empirical_log(ev_spsa,
                                   iters=800, nsamp=800,  # 完整训练
                                   warmup_iters=40, inner_steps=5,
                                   spsa_c_th=0.10, spsa_c_ph=0.10, spsa_c_al=0.05,
                                   spsa_exp=0.101,
                                   lr_theta=0.02, lr_phi=0.02, lr_alpha=0.10,
                                   tau0=0.05, tau_exp=0.5,
                                   nsamp_log=200,
                                   seed_train=7, seed_log=800,
                                   log_interval=20)

print("SPSA训练完成！绘制结果...")

# 绘制结果
plot_fig3b_spsa(res_spsa['logs'])

k_spsa = n_spsa // 2
codes_array_spsa = np.array([format(i, f'0{k_spsa}b') for i in range(1 << k_spsa)])[::-1]
plot_fig3c_from_results(ev_spsa, codes_array_spsa, res_spsa['theta'], res_spsa['phi'])
plt.show()

print("\n" + "=" * 60)
print("所有方法训练完成！")
print("=" * 60)
```

```python
# ========== Fig3(d) 完整成果代码 ==========
import sys
import os
sys.path.insert(0, 'Fig3')

import numpy as np
import matplotlib.pyplot as plt
import mindspore as ms

# MindSpore context
ms.set_context(mode=ms.PYNATIVE_MODE, device_target="CPU")

# 导入fig3d函数
from fig3d_functions import run_vqe_experiment, run_sh_vqe_experiment, plot_fig3d

print("=" * 60)
print("Fig3(d): VQE vs SH-VQE 比较")
print("=" * 60)

# ========== 实验配置 ==========
# 超参数配置
QUANTUM_NUMS = [8, 10, 12, 14, 16]
RESTARTS = 5
BASE_SEED = 21
STEPS = [100, 150, 200, 250, 300]  # 不同qubit对应不同步数
LR = 0.05
WARM_START = False
INIT_NOISE_SCALE = 0.01
VERBOSE_TRAIN = False

# depth配置
depths_config = {
    8: (2, 4, 6),
    10: (2, 4, 6),
    12: (2, 4, 6, 8),
    14: (2, 4, 6, 8, 10, 12, 14),
    16: (2, 4, 6, 8, 10, 12, 14, 16),
}

# ========== 运行实验 ==========
results_vqe = {}
results_sh = {}

for idx, n in enumerate(QUANTUM_NUMS):
    print(f"\n{'='*60}")
    print(f"开始 n={n} 的实验")
    print(f"{'='*60}")
    
    depths = depths_config[n]
    STEPS_current = STEPS[idx]
    
    # 运行VQE实验
    vqe_results, vqe_fid = run_vqe_experiment(
        n=n,
        depths=depths,
        pbc_brick=True,
        RESTARTS=RESTARTS,
        BASE_SEED=BASE_SEED,
        STEPS=STEPS_current,
        LR=LR,
        WARM_START=WARM_START,
        INIT_NOISE_SCALE=INIT_NOISE_SCALE,
        VERBOSE_TRAIN=VERBOSE_TRAIN
    )
    
    # 运行SH-VQE实验
    sh_results, sh_fid = run_sh_vqe_experiment(
        n=n,
        depths=depths,
        pbc_brick=True,
        RESTARTS=RESTARTS,
        BASE_SEED=BASE_SEED,
        STEPS=STEPS_current,
        LR=LR,
        WARM_START=WARM_START,
        INIT_NOISE_SCALE=INIT_NOISE_SCALE,
        VERBOSE_TRAIN=VERBOSE_TRAIN
    )
    
    results_vqe[n] = vqe_fid
    results_sh[n] = sh_fid
    
    print(f"n={n} 完成: VQE fidelities = {vqe_fid}")
    print(f"n={n} 完成: SH-VQE fidelities = {sh_fid}")

# ========== 组织数据并绘图 ==========
print("\n" + "=" * 60)
print("绘制Fig3(d)...")
print("=" * 60)

data = {}
for n in QUANTUM_NUMS:
    data[n] = {
        "vqe": results_vqe[n],
        "sh": results_sh[n]
    }

# 绘图
fig = plot_fig3d(data, depths_config)
plt.show()

print("\n" + "=" * 60)
print("Fig3(d) 完成！")
print("=" * 60)
```
