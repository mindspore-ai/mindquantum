### Calculate HF and FCI energies

```python
from funcs import *
from mindquantum import *
import json

def meta_analize(file_path:str, point:int):
    """Calculate HF and FCI energies"""
    with open(file_path, "r", encoding="utf-8") as f:
        save_data = json.load(f)
    
    geo = save_data["geos"][point]
    hf, fci, _, _, _, = get_mole_info(geo)
    method = "energys"
    to_save(file_path, method, point, 'hf', hf)
    to_save(file_path, method, point, 'fci', fci)


file_path = "H4.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(5)):
    meta_analize(file_path, point)

file_path = "HF.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(5)):
    meta_analize(file_path, point)

file_path = "LiH.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(5)):
    meta_analize(file_path, point)

file_path = "H2O.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(5)):
    meta_analize(file_path, point)

file_path = "BeH2.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(5)):
    meta_analize(file_path, point)

file_path = "NH3.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(5)):
    meta_analize(file_path, point)
```

### run adapt-vqe 

```python
from funcs import *
from mindquantum import *
import json

def grad_adapt_analize(file_path:str, point:int, max_size:int, tol:float=1e-4):
    """adapt-vqe"""
    with open(file_path, "r", encoding="utf-8") as f:
        save_data = json.load(f)
    
    geo = save_data["geos"][point]
    
    energys, errors, \
    params, mcounts, size_his = run_adapt_vqe(geo, max_size, tol=tol)
    method = "grad_adapt"
    to_save(file_path, method, point, 'energys', list(energys))
    to_save(file_path, method, point, 'errors', list(errors))
    to_save(file_path, method, point, 'mcounts', list(mcounts))
    to_save(file_path, method, point, 'params', list(params))
    to_save(file_path, method, point, 'size_his', size_his)

tol = 1e-4
max_size = 120
file_paths = ["H4.txt", "HF.txt", "LiH.txt", "H2O.txt", "BeH2.txt", "NH3.txt",]

for file_path in file_paths:
    print(f"{file_path}, grad_adapt, epsilon={tol}")
    for point in tqdm(range(5)):
        grad_adapt_analize(file_path, point, max_size, tol)
```
