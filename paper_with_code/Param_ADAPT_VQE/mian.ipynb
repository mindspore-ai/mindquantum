### Calculate HF and FCI energies

```python

def meta_analize(file_path:str, point:int):
    """Calculate HF and FCI energies"""
    with open(file_path, "r", encoding="utf-8") as f:
        save_data = json.load(f)
    
    geo = save_data["geos"][point]
    
    hf, fci, _, _, _, = get_mole_info(geo)
    method = "energys"
    to_save(file_path, method, point, 'hf', hf)
    to_save(file_path, method, point, 'fci', fci)


file_path = "H4.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(11)):
    meta_analize(file_path, point)

file_path = "H2O.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(11)):
    meta_analize(file_path, point)

file_path = "BeH2.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(11)):
    meta_analize(file_path, point)

file_path = "HF.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(11)):
    meta_analize(file_path, point)

file_path = "LiH.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(11)):
    meta_analize(file_path, point)

file_path = "NH3.txt"
print(f"{file_path} meta_analize")
for point in tqdm(range(11)):
    meta_analize(file_path, point)
```

### run adapt-vqe 

```python
from Param_funcs import *
from mindquantum import *
import json

def adapt_analize(file_path:str, point:int, max_iter:int, tol:float=1e-4):
    """adapt-vqe 综合分析"""
    with open(file_path, "r", encoding="utf-8") as f:
        save_data = json.load(f)
    
    geo = save_data["geos"][point]
    
    energys, errors, \
    params, mcounts = run_adapt_vqe(geo, max_iter, tol=tol)
    method = "grad_adapt"
    to_save(file_path, method, point, 'energys', list(energys))
    to_save(file_path, method, point, 'errors', list(errors))
    to_save(file_path, method, point, 'mcounts', list(mcounts))
    to_save(file_path, method, point, 'params', list(params))


tol = 1e-4
max_iter = 120
file_path = "H4.txt"
print(f"{file_path}, grad_adapt, epsilon={tol}")
for point in tqdm(range(11)):
    adapt_analize(file_path, point, max_iter, tol)

file_path = "H2O.txt"
print(f"{file_path}, grad_adapt, epsilon={tol}")
for point in tqdm(range(11)):
    adapt_analize(file_path, point, max_iter, tol)

file_path = "BeH2.txt"
print(f"{file_path}, grad_adapt, epsilon={tol}")
for point in tqdm(range(11)):
    adapt_analize(file_path, point, max_iter, tol)

file_path = "HF.txt"
print(f"{file_path}, grad_adapt, epsilon={tol}")
for point in tqdm(range(11)):
    adapt_analize(file_path, point, max_iter, tol)

file_path = "LiH.txt"
print(f"{file_path}, grad_adapt, epsilon={tol}")
for point in tqdm(range(11)):
    adapt_analize(file_path, point, max_iter, tol)

file_path = "NH3.txt"
print(f"{file_path}, grad_adapt, epsilon={tol}")
for point in tqdm(range(11)):
    adapt_analize(file_path, point, max_iter, tol)
```

### run param-adapt-vqe

```python
from Param_funcs import *
from mindquantum import *
import json


def param_analize(file_path:str, point:int, period:int, max_iter:int, tol:float=1e-4):
    """Comprehensive analysis"""
    with open(file_path, "r", encoding="utf-8") as f:
        save_data = json.load(f)
    
    geo = save_data["geos"][point]
    
    energys, errors, \
    params, mcounts = run_adapt_vqe_param(geo, max_iter=max_iter, tol=tol, period=period)
    method = f"param_adapt_{period}"
    to_save(file_path, method, point, 'energys', list(energys))
    to_save(file_path, method, point, 'errors', list(errors))
    to_save(file_path, method, point, 'mcounts', list(mcounts))
    to_save(file_path, method, point, 'params', list(params))


tol = 1e-4
max_iter = 120
for period in [1]:
    file_path = "H4.txt"
    print(f"{file_path}, param_adapt_{period}, epsilon={tol}")
    for point in tqdm(range(11)):
        param_analize(file_path, point, period, max_iter, tol)

    file_path = "H2O.txt"
    print(f"{file_path}, param_adapt_{period}, epsilon={tol}")
    for point in tqdm(range(11)):
        param_analize(file_path, point, period, max_iter, tol)

    file_path = "BeH2.txt"
    print(f"{file_path}, param_adapt_{period}, epsilon={tol}")
    for point in tqdm(range(11)):
        param_analize(file_path, point, period, max_iter, tol)

    file_path = "HF.txt"
    print(f"{file_path}, param_adapt_{period}, epsilon={tol}")
    for point in tqdm(range(11)):
        param_analize(file_path, point, period, max_iter, tol)

    file_path = "LiH.txt"
    print(f"{file_path}, param_adapt_{period}, epsilon={tol}")
    for point in tqdm(range(11)):
        param_analize(file_path, point, period, max_iter, tol)

    file_path = "NH3.txt"
    print(f"{file_path}, param_adapt_{period}, epsilon={tol}")
    for point in tqdm(range(11)):
        param_analize(file_path, point, period, max_iter, tol)
```
